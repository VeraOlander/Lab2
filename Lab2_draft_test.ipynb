```python
from numbers import Number
from validate_coordinates import validate_coordinates
# class Shape to represent the center position of the object

class Shape:
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y
  
    @property
    def x(self):
        return self._x
    
    @x.setter
    def x(self, x):
        validate_coordinates(x)        
        self._x = x


    @property
    def y(self):
        return self._y
    
    @y.setter
    def y(self, y):
        validate_coordinates(y)     
        self._y = y

    def __repr__(self):
        return f"Coordinates: x = {self.x}, y = {self.y}"
    

shape1 = Shape(1,0)
shape1
```

```python
shape2 = Shape(5,-4.7)
shape2
```

```python
from validate_parameters import validate_parameters
import math

class Circle(Shape):
    def __init__(self, x: Number, y: Number, radius: Number):
        super().__init__(x, y)
        self.radius=radius

    @property
    def radius(self):
        return self._radius
    
    @radius.setter
    def radius(self, radius):
        validate_parameters(radius)        
        self._radius = radius

    @property
    def area(self):
        return math.pi * (self._radius ** 2)
    
    @property
    def circumference(self):
        return 2 * math.pi * self._radius

    def __repr__(self):
        return f"This is a circle with a radius = {self.radius} and coordinates: x = {self.x}, y = {self.y}"
```

```python
circle1 = Circle(-2,9,10)
circle1
circle1.area
circle1.circumference
```
