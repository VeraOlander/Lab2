### Sphere class: manual testing

```python
from sphere import Sphere

sphere1 = Sphere(8,2,-2)
print(sphere1)
sphere1.translate(-3,-3)
print(sphere1)

try:
    sphere2 = Sphere(-9,7,7)
except ValueError as err:
    print(err)

sphere2 = Sphere(9,7,7)
print(sphere2)

print(sphere1==sphere2)
print(sphere1<sphere2)
print(sphere1>=sphere2)

print(sphere1.volume)
print(sphere2.volume)

print(sphere1.circumference)
print(sphere2.circumference)

print(sphere1.surface_area)
print(sphere2.surface_area)

```
